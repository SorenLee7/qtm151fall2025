## 5) Randomly assign each row into two groups
Add a helper column `random_var` with random numbers (Uniform[0,1)), then split into two DataFrames by a threshold.

```python
np.random.seed(SEED)
df["random_var"] = np.random.rand(len(df))

# Example threshold at 0.5 — adjust if needed
groupA = df.loc[df["random_var"] < 0.5].copy()
groupB = df.loc[df["random_var"] >= 0.5].copy()

print("Group A size:", len(groupA))
print("Group B size:", len(groupB))
```

## 6) Side-by-side scatter plots of the two groups
Use `matplotlib.pyplot.subplots` to put two scatter plots next to each other (Group A on the left, Group B on the right).

```python
# TODO: Set the x and y columns to plot from your DataFrame
x_col = "<XCOL>"
y_col = "<YCOL>"

fig, axes = plt.subplots(1, 2, figsize=(10, 4), constrained_layout=True)
axes[0].scatter(groupA[x_col], groupA[y_col])
axes[0].set_title("Group A")
axes[0].set_xlabel(x_col)
axes[0].set_ylabel(y_col)

axes[1].scatter(groupB[x_col], groupB[y_col])
axes[1].set_title("Group B")
axes[1].set_xlabel(x_col)
axes[1].set_ylabel(y_col)

plt.show()
```
