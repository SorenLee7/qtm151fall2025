
# Practice Quiz — Pandas Essentials (Solved)

This notebook mirrors six common Pandas tasks with concise, chainable solutions.

Datasets used:
- `main.csv` — synthetic sample with `id, group, city, value, score, status`
- `lookup.csv` — city to region map


````

import pandas as pd

main_path = r"/mnt/data/main.csv"
lookup_path = r"/mnt/data/lookup.csv"

df = pd.read_csv(main_path)
lookup = pd.read_csv(lookup_path)

df.head()
````


## 1) Replace three values in a single column using one `.replace` call

**Task:** In column `status`, replace `old_red → red`, `old_green → green`, and `old_blue → blue` using `.replace` once.


````

# Solution
df_q1 = df.copy()
df_q1["status"] = df_q1["status"].replace({"old_red": "red", "old_green": "green", "old_blue": "blue"})
df_q1["status"].value_counts()
````


## 2) Filter → group → aggregate → filter again, all in one chain

**Task:** Subset rows where `score >= 50`, then group by `group` and compute `value_mean` and `score_sum`.
Finally, return only groups with `score_sum >= 150`. Do this in **one line** using chaining.


````

# Solution (one chain)
res_q2 = (
    df.loc[lambda d: d["score"] >= 50]
      .groupby("group", as_index=False)
      .agg(value_mean=("value","mean"), score_sum=("score","sum"))
      .loc[lambda g: g["score_sum"] >= 150]
)

res_q2
````


## 3) Recode a numeric variable into categorical intervals

**Task:** Convert `value` into a categorical `value_band` based on intervals:
- [0, 40): `"low"`
- [40, 70): `"mid"`
- [70, ∞): `"high"`


````

# Solution
bins = [0, 40, 70, float("inf")]
labels = ["low", "mid", "high"]
df_q3 = df.copy()
df_q3["value_band"] = pd.cut(df_q3["value"], bins=bins, labels=labels, right=False)
df_q3[["value","value_band"]].head(12)
````


## 4) Group → aggregate → sort by one aggregated statistic in descending order in one chain

**Task:** Group by `city`. Aggregate `n = count`, and `score_mean = mean score`. Sort by `score_mean` descending.


````

# Solution (one chain)
res_q4 = (
    df.groupby("city", as_index=False)
      .agg(n=("id","count"), score_mean=("score","mean"))
      .sort_values("score_mean", ascending=False)
)

res_q4
````


## 5) Rename exactly one column

**Task:** Rename `value` → `value_points` and leave all other columns unchanged.


````

# Solution
df_q5 = df.rename(columns={"value": "value_points"}).head()
df_q5
````


## 6) Merge two DataFrames, adding one new column from the secondary table

**Task:** Merge `df` with `lookup` on `city` so that `region` is added to the main table.


````

# Solution
df_q6 = df.merge(lookup, on="city", how="left")
df_q6.head()
````


## One-liner patterns to memorize

- Replace map:
```python
df["col"] = df["col"].replace({old1:new1, old2:new2, old3:new3})
```
- Filter → group → aggregate → filter:
```python
(df.loc[lambda d: cond(d)]
   .groupby(keys, as_index=False)
   .agg(...)
   .loc[lambda g: cond2(g)])
```
- Numeric → categorical bins:
```python
df["cat"] = pd.cut(df["num"], bins=[...], labels=[...], right=False)
```
- Group → aggregate → sort:
```python
(df.groupby(keys, as_index=False)
   .agg(...)
   .sort_values(col, ascending=False))
```
- Rename single column:
```python
df = df.rename(columns={{"old":"new"}})
```
- Left merge to add a lookup:
```python
df = df.merge(lookup, on="key", how="left")
```
