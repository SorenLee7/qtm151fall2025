## 1) Fix a Wrong Plot for a Quadratic Model

**Given on the quiz:** A previous cell correctly computed quadratic coefficients for  
\(\hat{y} = a x^2 + b x + c\). Your task is to **fix the plotting cell**.

**Common mistakes to look for (and fix):**
- Plotting `yhat` (predictions) against the **row index** rather than the actual \(x\) values.
- Not **sorting** \(x\) before drawing the fitted curve (causes a zig-zag line).
- Swapping axes by mistake (e.g., `plt.plot(yhat, x)` instead of `plt.plot(x, yhat)`).
- Accidentally recomputing predictions on a **mismatched** array (e.g., different length/order than `x` used in the fit).
- Forgetting labels/legend (less critical, but clarity helps).

Below is the **correct pattern**. Replace the placeholders to match the quiz variable names.


```
# --- CORRECT PLOTTING TEMPLATE (fill placeholders) ---
# Assumptions: Previous cell already did something like:
# a, b, c = np.polyfit(<X_ARRAY>, <Y_ARRAY>, 2)

# 1) Scatter the original data
plt.figure()
plt.scatter(<X_ARRAY>, <Y_ARRAY>, label="<DATA_LABEL>")  # e.g., "data"

# 2) Sort x for a smooth line
_order = np.argsort(<X_ARRAY>)
_x_sorted = <X_ARRAY>[_order]
# Use the already computed coefficients a, b, c from the previous (correct) cell
_yhat_sorted = a*_x_sorted**2 + b*_x_sorted + c

# 3) Plot model prediction against the **sorted x**
plt.plot(_x_sorted, _yhat_sorted, label="quadratic fit")

# 4) Labels, legend
plt.xlabel("<X_LABEL>")  # e.g., "x"
plt.ylabel("<Y_LABEL>")  # e.g., "y"
plt.title("<TITLE>")     # e.g., "Quadratic Model: data vs fitted curve"
plt.legend()
plt.show()

# --- Notes on likely wrong quiz code and how to fix ---
# If you see code like: plt.plot(range(len(yhat)), yhat)  -> Replace x-axis with the actual x (sorted)
# If you see: plt.plot(yhat, x)                             -> Flip to plt.plot(x, yhat)
# If you see: yhat = a*x**2 + b*x + c where x is different from fit data -> make sure x matches the data used for fit
# If you see no sorting: add argsort sorting before plotting the line
```

## 3) FX Rates — Non-Dimensional % Growth

**Goal:** Read a CSV with **EUR→USD** and **GBP→USD** series. Compute % growth relative to the **first** observation:

\[%Growth(t) = (x_t / x_{t_0} - 1) \times 100\]

Then plot both % growth series together (use either `DataFrame.plot()` or `plt.plot`). Fill placeholders for file path and column names.


```
# --- TEMPLATE: % growth from first value ---
FX_CSV = "<CSV_PATH_TO_FX>"
DATE_COL = "<DATE_COL>"            # e.g., "date"
EUR_COL  = "<EURUSD_COL_NAME>"     # e.g., "EURUSD"
GBP_COL  = "<GBPUSD_COL_NAME>"     # e.g., "GBPUSD"

df_fx = pd.read_csv(FX_CSV, parse_dates=[DATE_COL]).sort_values(DATE_COL).set_index(DATE_COL)

# Percentage growth columns
df_fx[EUR_COL + "_pct_growth"] = (df_fx[EUR_COL] / df_fx[EUR_COL].iloc[0] - 1.0) * 100.0
df_fx[GBP_COL + "_pct_growth"] = (df_fx[GBP_COL] / df_fx[GBP_COL].iloc[0] - 1.0) * 100.0

# Plot both
ax = df_fx[[EUR_COL + "_pct_growth", GBP_COL + "_pct_growth"]].plot()
ax.set_xlabel("<TIME_AXIS_LABEL>")
ax.set_ylabel("% growth from first day")
ax.set_title("<TITLE>")  # e.g., "EURUSD vs GBPUSD — % Growth")
plt.show()
```

## 6) Finite Differences with `.diff()` — Check \(\frac{d}{dt}\sin t = \cos t\)

**Goal:** Build a DataFrame with columns:
- `time`: evenly spaced between \(t_0\) and \(t_1\) (e.g., \(0\) to \(4\pi\))
- `sin`: \(\sin(\text{time})\)
- `cos`: \(\cos(\text{time})\)

Then use `.diff()` to approximate the derivative of `sin`:
\[\text{d\_sin\_dt}_i \approx \frac{\sin(t_i) - \sin(t_{i-1})}{t_i - t_{i-1}}.\]

Plot **both** `(time, cos)` and `(time, d_sin_dt)` together to show they are close.

Fill placeholders for the time grid details if specified on your quiz.


```
# --- TEMPLATE: finite difference derivative ---
N_POINTS = <N_POINTS>        # e.g., 400
T0       = <T_START>         # e.g., 0.0
T1       = <T_END>           # e.g., 4*np.pi

time = np.linspace(T0, T1, N_POINTS)
sin_t = np.sin(time)
cos_t = np.cos(time)

df_trig = pd.DataFrame({"time": time, "sin": sin_t, "cos": cos_t})

# Finite difference derivative via .diff()
df_trig["d_sin_dt"] = df_trig["sin"].diff() / df_trig["time"].diff()

plt.figure()
plt.plot(df_trig["time"], df_trig["cos"], label="cos(t)")
plt.plot(df_trig["time"], df_trig["d_sin_dt"], label="d(sin)/dt (finite diff)")
plt.xlabel("time")
plt.ylabel("value")
plt.title("cos(t) vs. finite-difference derivative of sin(t)")
plt.legend()
plt.show()
```
